## Part A: Vectors

<strong>A1</strong> \[2 pt\] Write a function called <strong>vector_ops</strong> that takes as inputs a scalar $c$ and two lists $v$ and $w$.<br>
It should return a dictionary with eight key-value pairs: 
* <strong>shape</strong>: the shape of $v$ (as a tuple) 
* <strong>scale</strong>: $v$ scaled by $c$ (as a numpy array) 
* <strong>addition</strong>: $v$ plus $w$ (as a numpy array) 
* <strong>subtraction</strong>: $v$ minus $w$ (as a numpy array) 
* <strong>dot</strong>: the dot product of $v$ and $w$ (as a numpy float) 
* <strong>length</strong>: the norm of $v$ (as a numpy float) 
* <strong>as_one_long_vector</strong>: v and w stacked horizontally (as a numpy array)
* <strong>as_rows_in_a_matrix</strong>: v and w stacked vertically (as a numpy array)

```python
def vector_ops(c,v,w):
    # YOUR CODE HERE
    v = np.array(v)
    w = np.array(w)

    result = {
        'shape' : v.shape,
        'scale' : c * v,
        'addition' : v + w,
        'subtraction' : v - w,
        'dot' : v @ w,
        'length' : np.linalg.norm(v),
        #'length' : np.sqrt(v[0]**2 + v[1]**2),
        'as_one_long_vector' :  np.hstack((v, w)),
        'as_rows_in_a_matrix' : np.vstack((v, w))
    }
    return result
```

Your function should return the boolean True if the vectors are linearly independent and the boolean False otherwise. 

```python
def indep_or_not(vectors):
    # YOUR CODE HERE
    mat = np.column_stack(vectors)
    rank = np.linalg.matrix_rank(mat)
    #print(np.linalg.matrix_rank(mat)) 
    return rank == len(vectors)
```

Your function should return the boolean True if the vectors span $\mathbb{R}^{m}$ and the boolean False otherwise. 

```python
def span_or_not(vectors):
    # YOUR CODE HERE
    mat = np.column_stack(vectors)
    #display(mat)
    rank = np.linalg.matrix_rank(mat)
    #print(rank)
    return rank == mat.shape[0]
```

<strong>C1</strong> \[2 pt\] Write a function called <strong>matrix_ops</strong> that has three inputs: a scalar $c$ (a float) and two matrices $A$ and $B$ (numpy arrays).<br> 
It should return a dictionary with seven key-value pairs:
* <strong>shape</strong>: the shape of $A$ (as a tuple)
* <strong>scale</strong>: $A$ scaled by $c$ (as a numpy array)
* <strong>sum</strong>: $A$ plus $B$ (as a numpy array)
* <strong>difference</strong>: $A$ minus $B$ (as a numpy array)
* <strong>elementwise_multiplication</strong>: elementwise multiplication of $A$ and $B$ (as a numpy array) 
* <strong>matrix_multiplication</strong>: matrix multiplication of $A$ and $B$ (as a numpy array)
* <strong>transpose</strong>: $A$ transposed (as a numpy array)

**If the shapes of $A$ and $B$ are such that a key's value is not possible, then make its value the string "not possible".**

```python
def check_comp(A, B):
    return {
        'add': A.shape == B.shape,
        'sub': A.shape == B.shape,
        'mult': A.shape[-1] == B.shape[0]
    }
```

```python
def matrix_ops(c,A,B):
    # YOUR CODE HERE
    A = np.array(A)
    B = np.array(B)

    comp = check_comp(A, B)

    result = {
        'shape' : A.shape,
        'scale' : c * A,
        'sum': A + B if comp['add'] else 'not possible',
        'difference' :  A - B if comp['sub'] else 'not possible',
        'elementwise_multiplication' : A * B if comp['add'] else 'not possible',
        'matrix_multiplication' : A @ B if comp['mult'] else 'not possible',
        'transpose' :  A.T,
    }
    return result 
```

```python
def matrix_inverse(square_matrix):
    # YOUR CODE HERE
    square_matrix = np.array(square_matrix)
    
    if (square_matrix.shape[0] == square_matrix.shape[1]) and \
       (square_matrix.shape[0] == np.linalg.matrix_rank(square_matrix)):
        return np.linalg.inv(square_matrix)
    else:
        return 0
```

## Part E: Matrices as Linear Mappings

<strong>E1</strong> \[2 pt\] Write a function called <strong>matrix_classifier</strong> that has one input called Matrix1 represented as a numpy array. The function should classify the matrix as one of (i) one-to-one, (ii) onto, (iii) both (i.e. invertible), or (iv) neither. It should return the classification represented as a string (i.e. "one-to-one", "onto", "invertible", or "neither").

```python
def matrix_classifier(Matrix1):
    # YOUR CODE HERE
    rank = np.linalg.matrix_rank(Matrix1)
    if (rank == Matrix1.shape[0]) and (rank == Matrix1.shape[1]):
        return 'invertible'
    elif rank == Matrix1.shape[0]:
        return 'onto' #spans
    elif rank == Matrix1.shape[1]:
        return 'one-to-one' #independent
    else:
        return 'neither'
        
```
