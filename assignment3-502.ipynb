<strong>A6</strong> \[3 pt\] The experiment also wants the mean age in the treatment and control groups to be similar. Suppose that the ages of the 16 participants are stored in the variable ages.

$ages = [74, 76, 26, 71, 23, 31, 43, 45, 65, 33, 34, 27, 56, 29, 52, 21]$

If the experiment still wants 8 people in each group, what fraction of all possible treatment-control splits will be such that the mean age in the two groups are LESS THAN five years of each other? Store your answer in the variable 
<strong>frac_groups_with_similar_mean_age</strong> as a float. 


<strong>Note</strong>: This question builds on the narrative from the last two questions, but for this question, don't worry at all about gender.

<strong>Hint</strong>: Use itertools and a for-loop.

```python
ages = [74,76,26,71,23,31,43,45,65,33,34,27,56,29,52,21]

frac_groups_with_similar_mean_age = None
# YOUR CODE HERE
count_similar = 0
results = list(itertools.combinations(ages, 8))
for group in results:
    treatment_mean = np.mean(group)
    control_mean = np.mean([age for age in ages if age not in group])
    difference = abs(treatment_mean - control_mean)
    if difference < 5:
        count_similar += 1
    else:
        continue

frac_groups_with_similar_mean_age = float(count_similar / len(results))
frac_groups_with_similar_mean_age        
```

<strong>B1</strong> \[2 pt\] Suppose that the random variable $X$ can take on any integer value except 10. The probability that it takes on the value $x$ is given by $p(x) = k (\frac{1}{2})^{|x-10|}$. What value of $k$ make this a valid pmf? What is the expected value of $X$? Store your answers in the variables <strong>k_for_valid_pmf</strong> and <strong>expected_value_X</strong> as floats.

```python
k_for_valid_pmf = None
expected_value_X = None
# YOUR CODE HERE
total = 0
for x in range(-1000, 1000):
    if x != 10:
        total += (1/2) ** abs(x - 10)

k_for_valid_pmf = 1 / total

print(k_for_valid_pmf)

expected_value_X = 0
for x in range(-1000, 1000):
    if x != 10:
        expected_value_X += x * (k_for_valid_pmf) * ((1/2) ** abs(x - 10))

print(expected_value_X)    
```
