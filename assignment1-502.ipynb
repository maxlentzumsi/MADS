<strong>A2</strong> \[2 pt\] Write a function <strong>frac_divisible_x_or_y_not_both</strong> that takes as inputs positive integers $L$, $U$, $x$, and $y$ that returns the answer to the question ``What fraction of the integers from $L$ to $U$ (inclusive) are divisible by $x$ or $y$ but not both?"

```python
def frac_divisible_x_or_y_not_both(L, U, x, y):
    # YOUR CODE HERE
    divisible_x = len([i for i in range(L, U+1) if i%x == 0])
    divisible_y = len([i for i in range(L, U+1) if i%y == 0])
    divisible_both = len([i for i in range(L, U+1) if i%x == 0 and i%y==0])

    frac_divisible_x_or_y_not_both = (divisible_x + divisible_y - 2 * divisible_both) / (1 + U - L)
    return frac_divisible_x_or_y_not_both
```
