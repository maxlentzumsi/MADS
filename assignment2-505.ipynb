````python
def proportion_of_education():
    # your code goes here
    # YOUR CODE HERE
    # column EDUC1:
    # less than high school = 1
    # high school = 2
    # more than high school but not college = 3
    # college = 4
    
    file = open("NISPUF17.csv", "r")
    df = pd.read_csv(file)
    #df = pd.read_csv("NISPUF17.csv")
    
    total_children = df["CHILDNM"].count()
    less_than_high_school = (df[df["EDUC1"] == 1]["CHILDNM"].count() / total_children)
    high_school = (df[df["EDUC1"] == 2]["CHILDNM"].count() / total_children)
    more_than_high_school_but_not_college = (df[df["EDUC1"] == 3]["CHILDNM"].count() / total_children)
    college = (df[df["EDUC1"] == 4]["CHILDNM"].count() / total_children)
    Total_percentage = less_than_high_school + high_school + more_than_high_school_but_not_college + college
    #print(f"""
    #Total children: {total_children}
    #Less than high school: {less_than_high_school}
    #High school: {high_school}
    #More than high school but not college: {more_than_high_school_but_not_college}
    #College: {college}
    #Total percentage = {Total_percentage}
    #""")
    d = {"less than high school": less_than_high_school,
    "high school": high_school,
    "more than high school but not college": more_than_high_school_but_not_college,
    "college": college}   
    return d
    
    #raise NotImplementedError()
````

## Question 2: chickenpox_by_sex()
It would be interesting to see if there is any evidence of a link between vaccine effectiveness and sex of the child. Calculate the ratio of the number of children who contracted chickenpox but were vaccinated against it (at least one varicella dose) versus those who were vaccinated but did not contract chicken pox. Return results by sex. 

*This function should return a dictionary in the form of (use the correct numbers):* 
```
    {"male":0.2,
    "female":0.4}
```

Note: To aid in verification, the `chickenpox_by_sex()['female']` value the autograder is looking for starts with the digits `0.00779`.

````python
def chickenpox_by_sex():
    # YOUR CODE HERE
    # SEX - 1 = MALE | 2 = FEMALE
    #P_NUMVRC - total number of varicella doses
    #HAD_CPOX - 1 = YES | 2 = NO

    file = open("NISPUF17.csv", "r")
    df = pd.read_csv(file)
    #print(df[["SEX", "P_NUMVRC", "HAD_CPOX"]].head())
    
    #total_children = df["CHILDNM"].count()

    #Calculate male
    total_children_male_contracted = ((df["SEX"] == 1) & (df["HAD_CPOX"] == 1)).sum()
    total_children_male_vax_contract = ((df["SEX"] == 1) & (df["HAD_CPOX"] == 1) & (df["P_NUMVRC"] >= 1)).sum()
    total_children_male_vax_non_contract = ((df["SEX"] == 1) & (df["HAD_CPOX"] == 2) & (df["P_NUMVRC"] >= 1)).sum()

    ratio_male = total_children_male_vax_contract / total_children_male_vax_non_contract

    #Calculate female
    total_children_female_contracted = ((df["SEX"] == 2) & (df["HAD_CPOX"] == 1)).sum()
    total_children_female_vax_contract = ((df["SEX"] == 2) & (df["HAD_CPOX"] == 1) & (df["P_NUMVRC"] >= 1)).sum()
    total_children_female_vax_non_contract = ((df["SEX"] == 2) & (df["HAD_CPOX"] == 2) & (df["P_NUMVRC"] >= 1)).sum()

    ratio_female = total_children_female_vax_contract / total_children_female_vax_non_contract
    

    return {
        "male": ratio_male
        ,"female": ratio_female
    }
````

A correlation is a statistical relationship between two variables. If we wanted to know if vaccines work, we might look at the correlation between the use of the vaccine and whether it results in prevention of the infection or disease [1]. In this question, you are to see if there is a correlation between having had the chicken pox and the number of chickenpox vaccine doses given (varicella).

Some notes on interpreting the answer. The `had_chickenpox_column` is either `1` (for yes) or `2` (for no), and the `num_chickenpox_vaccine_column` is the number of doses a child has been given of the varicella vaccine. A positive correlation (e.g., `corr > 0`) means that an increase in `had_chickenpox_column` (which means more no's) would also increase the values of `num_chickenpox_vaccine_column` (which means more doses of vaccine). If there is a negative correlation (e.g., `corr < 0`), it indicates that having had chickenpox is related to an increase in the number of vaccine doses. 

Also, `pval` is the probability that we observe a correlation between `had_chickenpox_column` and `num_chickenpox_vaccine_column` which is greater than or equal to a particular value occurred by chance. A small `pval` means that the observed correlation is highly unlikely to occur by chance. In this case, `pval` should be very small (will end in `e-18` indicating a very small number).
